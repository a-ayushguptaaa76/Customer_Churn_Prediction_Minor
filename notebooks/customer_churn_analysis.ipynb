## 1. Dataset overview

The baseline dataset contains 10,000 customer records, 11 predictors, and a binary `churn` target. The recorded class distribution was 79.63% non-churn and 20.37% churn.


```python
import pandas as pd
from src.features import add_features

df = pd.read_csv("data/customer_data.csv")
display(df.head())
print("shape:", df.shape)
print(df["churn"].value_counts(normalize=True).rename("proportion"))
```

## 3. Reproducible training

From the repository root run:

python -m src.train --data data/customer_data.csv

The command compares three classifiers with stratified 5-fold ROC-AUC, evaluates a held-out test set, and writes `reports/metrics.json` plus `models/churn_pipeline.joblib`.


```python
import json
from pathlib import Path

report_path = Path("reports/metrics.json")
if report_path.exists():
    report = json.loads(report_path.read_text())
    display(pd.DataFrame(report["results"]))
else:
    print("Run the training command first to generate reports/metrics.json")
```

## 4. Historical baseline

The earlier experiment recorded mean 5-fold ROC-AUC values of 0.7877 (Logistic Regression), 0.8486 (Random Forest), 0.8628 (Gradient Boosting), 0.8462 (AdaBoost), and 0.8351 (SVC). The held-out Gradient Boosting result was 0.8692 ROC-AUC and 0.8680 accuracy.

Those figures are retained as historical baseline evidence rather than presented as new run results.
